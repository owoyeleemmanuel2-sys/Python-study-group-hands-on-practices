```python
# EX. 2
# Generating a payroll reciept automatically
#Variables
employee_name = "David Musa"
department = "finance"
basic_salary = 185000
bonus = 35000
# Calculating
gross_salary = basic_salary + bonus
tax = gross_salary *(7.5 / 100)
net_salary = gross_salary - tax
# Display Output
report = f'''==============================
     Payroll Reciept
==============================
Employee: {employee_name}
Department: {department}
Basic Salary: #{basic_salary}
Gross Salary: #{gross_salary}
Tax(7.5): #{tax}
Net Salary: #{net_salary}
==============================
'''
print(report)
```

```python
# Variables
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000.00
bonus = 35000.00
tax_rate = 0.075  # 7.5%

# Calculations
gross_salary = basic_salary + bonus
tax_amount = gross_salary * tax_rate
net_salary = gross_salary - tax_amount

# Display Output
print("==================================")
print("PAYROLL SUMMARY")
print("==================================")
print(f"Employee : {employee_name}")
print(f"Department : {department}")
print(f"Basic Salary : ₦{basic_salary:,.2f}")
print(f"Bonus : ₦{bonus:,.2f}")
print(f"Gross Salary : ₦{gross_salary:,.2f}")
print(f"Tax (7.5%) : ₦{tax_amount:,.2f}")
print(f"Net Salary : ₦{net_salary:,.2f}")
print("==================================")
```

```python
# EX. 3
# Online Store Reciept
# Variables
name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500
discount_allowed = quantity >= 2 #(10%)
# Calculation
total_cost = unit_price * quantity
if discount_allowed:
    discount = total_cost * 0.1
    final_payment = total_cost - discount
# Display Output

reciept = f'''==============================
     Online Store Reciept
==============================
Name: {name}
Product: {product}
Quantity: {quantity}
Unit Price: # {unit_price}
Total Cost: #{total_cost}
Discount(10%): #{discount}
Final Payment: #{final_payment}

Thank you for shopping with us!
==============================
'''
print(reciept)
```
